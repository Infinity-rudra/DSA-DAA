```
def insertion_sort(arr):
    """
    Sorts a list in-place using Insertion Sort and counts comparisons.

    Args:
        arr (list): The list of numbers to sort.

    Returns:
        int: The total number of key comparisons performed.
    """
    comparisons = 0
    # Start from the second element (index 1)
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1

        # Move elements of arr[0..i-1], that are greater than key,
        # to one position ahead of their current position
        while j >= 0:
            comparisons += 1  # Count the comparison
            if key < arr[j]:
                arr[j + 1] = arr[j]
                j -= 1
            else:
                # The element is in its correct place
                break

        arr[j + 1] = key

    return comparisons

# --- Main execution ---
if __name__ == "__main__":
    # Example array to test the sorting
    my_array = [12, 11, 13, 5, 6, 7, 2, 15, 1]

    print(f"Original array: {my_array}")

    # Create a copy to sort, so we can show the original
    arr_to_sort = list(my_array)

    # Call the sorting function
    total_comparisons = insertion_sort(arr_to_sort)

    print(f"Sorted array:   {arr_to_sort}")
    print(f"Total comparisons made: {total_comparisons}")

    # Another example
    already_sorted = [1, 2, 3, 4, 5]
    print(f"\nOriginal array: {already_sorted}")
    comparisons_sorted = insertion_sort(already_sorted)
    print(f"Sorted array:   {already_sorted}")
    print(f"Total comparisons made: {comparisons_sorted}")

    # Worst-case example
    reverse_sorted = [5, 4, 3, 2, 1]
    print(f"\nOriginal array: {reverse_sorted}")
    comparisons_reverse = insertion_sort(reverse_sorted)
    print(f"Sorted array:   {reverse_sorted}")
    print(f"Total comparisons made: {comparisons_reverse}")
```

```
def merge_sort(arr):
    """
    Sorts a list using Merge Sort and counts comparisons.
    This function calls a helper to perform the merge.

    Args:
        arr (list): The list of numbers to sort.

    Returns:
        tuple: A tuple containing the sorted list (a new list)
               and the total number of key comparisons.
    """
    comparisons = 0

    if len(arr) > 1:
        # Find the middle of the list
        mid = len(arr) // 2

        # Divide the list into two halves
        left_half = arr[:mid]
        right_half = arr[mid:]

        # Recursively sort both halves
        # The function returns (sorted_list, num_comparisons)
        sorted_left, comps_left = merge_sort(left_half)
        sorted_right, comps_right = merge_sort(right_half)

        # Add comparisons from recursive calls
        comparisons += comps_left
        comparisons += comps_right

        # --- Merge the two sorted halves ---
        i = j = k = 0
        merged_array = [0] * len(arr)

        # Copy data to temp lists sorted_left and sorted_right
        while i < len(sorted_left) and j < len(sorted_right):
            comparisons += 1  # This is the key comparison
            if sorted_left[i] < sorted_right[j]:
                merged_array[k] = sorted_left[i]
                i += 1
            else:
                merged_array[k] = sorted_right[j]
                j += 1
            k += 1

        # Check for any remaining elements in left_half
        while i < len(sorted_left):
            merged_array[k] = sorted_left[i]
            i += 1
            k += 1

        # Check for any remaining elements in right_half
        while j < len(sorted_right):
            merged_array[k] = sorted_right[j]
            j += 1
            k += 1

        return merged_array, comparisons

    else:
        # Base case: a list of 0 or 1 elements is already sorted
        return arr, 0

# --- Main execution ---
if __name__ == "__main__":
    # Example array to test the sorting
    my_array = [12, 11, 13, 5, 6, 7, 2, 15, 1]

    print(f"Original array: {my_array}")

    # Call the sorting function
    # Note: Merge Sort is not in-place, it returns a new sorted list
    sorted_array, total_comparisons = merge_sort(my_array)

    print(f"Sorted array:   {sorted_array}")
    print(f"Total comparisons made: {total_comparisons}")

    # Another example
    other_array = [38, 27, 43, 3, 9, 82, 10]
    print(f"\nOriginal array: {other_array}")
    sorted_array_2, total_comparisons_2 = merge_sort(other_array)
    print(f"Sorted array:   {sorted_array_2}")
    print(f"Total comparisons made: {total_comparisons_2}")
```

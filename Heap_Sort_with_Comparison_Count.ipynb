```
def heapify(arr, n, i, comparisons):
    """
    To heapify a subtree rooted at index i.
    n is size of heap.
    comparisons is a list [count] to store the comparison count.
    """
    largest = i  # Initialize largest as root
    left = 2 * i + 1
    right = 2 * i + 2

    # See if left child of root exists and is greater than root
    if left < n:
        comparisons[0] += 1
        if arr[i] < arr[left]:
            largest = left

    # See if right child of root exists and is greater than largest so far
    if right < n:
        comparisons[0] += 1
        if arr[largest] < arr[right]:
            largest = right

    # Change root, if needed
    if largest != i:
        arr[i], arr[largest] = arr[largest], arr[i]  # swap

        # Heapify the root.
        heapify(arr, n, largest, comparisons)

def heap_sort(arr):
    """
    Sorts a list in-place using Heap Sort and counts comparisons.

    Args:
        arr (list): The list of numbers to sort.

    Returns:
        int: The total number of key comparisons performed.
    """
    n = len(arr)
    comparisons = [0]  # Use a list to pass by reference

    # Build a maxheap.
    # We start from the last non-leaf node and go up.
    for i in range(n // 2 - 1, -1, -1):
        heapify(arr, n, i, comparisons)

    # One by one extract elements
    for i in range(n - 1, 0, -1):
        arr[i], arr[0] = arr[0], arr[i]  # swap
        heapify(arr, i, 0, comparisons)

    return comparisons[0]

# --- Main execution ---
if __name__ == "__main__":
    # Example array to test the sorting
    my_array = [12, 11, 13, 5, 6, 7, 2, 15, 1]

    print(f"Original array: {my_array}")

    # Create a copy to sort, so we can show the original
    arr_to_sort = list(my_array)

    # Call the sorting function
    total_comparisons = heap_sort(arr_to_sort)

    print(f"Sorted array:   {arr_to_sort}")
    print(f"Total comparisons made: {total_comparisons}")

    # Another example
    other_array = [38, 27, 43, 3, 9, 82, 10]
    print(f"\nOriginal array: {other_array}")
    arr_to_sort_2 = list(other_array)
    total_comparisons_2 = heap_sort(arr_to_sort_2)
    print(f"Sorted array:   {arr_to_sort_2}")
    print(f"Total comparisons made: {total_comparisons_2}")
```

```
def partition(arr, low, high, comparisons):
    """
    Helper function to find the partition position.
    Uses the last element as the pivot.
    comparisons is a list [count] to store the comparison count.
    """
    # Choose the rightmost element as pivot
    pivot = arr[high]

    # pointer for greater element
    i = low - 1

    # traverse through all elements
    # compare each element with pivot
    for j in range(low, high):
        comparisons[0] += 1
        if arr[j] <= pivot:
            # If element smaller than pivot is found
            # swap it with the greater element pointed by i
            i = i + 1

            # Swapping element at i with element at j
            arr[i], arr[j] = arr[j], arr[i]

    # Swap the pivot element with the greater element specified by i
    arr[i + 1], arr[high] = arr[high], arr[i + 1]

    # Return the position from where partition is done
    return i + 1

def quick_sort_recursive(arr, low, high, comparisons):
    """
    Recursive helper function for Quick Sort.
    """
    if low < high:
        # Find pivot element such that
        # element smaller than pivot are on the left
        # element greater than pivot are on the right
        pi = partition(arr, low, high, comparisons)

        # Recursive call on the left of pivot
        quick_sort_recursive(arr, low, pi - 1, comparisons)

        # Recursive call on the right of pivot
        quick_sort_recursive(arr, pi + 1, high, comparisons)

def quick_sort(arr):
    """
    Sorts a list in-place using Quick Sort and counts comparisons.

    Args:
        arr (list): The list of numbers to sort.

    Returns:
        int: The total number of key comparisons performed.
    """
    comparisons = [0]  # Use a list to pass by reference
    n = len(arr)

    quick_sort_recursive(arr, 0, n - 1, comparisons)

    return comparisons[0]

# --- Main execution ---
if __name__ == "__main__":
    # Example array to test the sorting
    my_array = [12, 11, 13, 5, 6, 7, 2, 15, 1]

    print(f"Original array: {my_array}")

    # Create a copy to sort, so we can show the original
    arr_to_sort = list(my_array)

    # Call the sorting function
    total_comparisons = quick_sort(arr_to_sort)

    print(f"Sorted array:   {arr_to_sort}")
    print(f"Total comparisons made: {total_comparisons}")

    # Another example
    other_array = [38, 27, 43, 3, 9, 82, 10]
    print(f"\nOriginal array: {other_array}")
    arr_to_sort_2 = list(other_array)
    total_comparisons_2 = quick_sort(arr_to_sort_2)
    print(f"Sorted array:   {arr_to_sort_2}")
    print(f"Total comparisons made: {total_comparisons_2}")
```

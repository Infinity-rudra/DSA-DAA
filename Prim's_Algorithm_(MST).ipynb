**Prim's Algorithm (MST)**

```
import sys

class Graph:
    def __init__(self, vertices):
        """
        Initializes the graph with a given number of vertices.
        Args:
            vertices (int): The number of vertices (nodes) in the graph.
        """
        self.V = vertices
        # Initialize an adjacency matrix with all 0s
        self.graph = [[0 for _ in range(vertices)] for _ in range(vertices)]

    def print_mst(self, parent):
        """
        Prints the edges of the constructed Minimum Spanning Tree (MST).
        Args:
            parent (list): A list where parent[i] stores the node
                           that connects 'i' to the MST.
        """
        print("Edge \tWeight")
        total_cost = 0
        for i in range(1, self.V):
            # parent[i] is the node, i is the destination
            # self.graph[i][parent[i]] is the weight of that edge
            weight = self.graph[i][parent[i]]
            print(f"{parent[i]} - {i}\t{weight}")
            total_cost += weight
        print(f"\nTotal Minimum Spanning Tree Cost: {total_cost}")

    def min_key(self, key, mst_set):
        """
        A helper function to find the vertex with the minimum key value,
        from the set of vertices not yet included in the MST.
        """
        # Initialize min value to infinity
        min_val = sys.maxsize
        min_index = -1

        for v in range(self.V):
            # Check if v is not in MST and its key is smaller
            if key[v] < min_val and not mst_set[v]:
                min_val = key[v]
                min_index = v
        return min_index

    def prim_mst(self):
        """
        Finds the Minimum Spanning Tree of the graph using Prim's algorithm.
        The graph is represented by the adjacency matrix self.graph.
        """
        # 'key' list stores the minimum edge weight to connect vertex i to the MST
        key = [sys.maxsize] * self.V

        # 'parent' list stores the constructed MST
        parent = [None] * self.V

        # 'mst_set' is a boolean list, mst_set[i] is True if vertex i is in the MST
        mst_set = [False] * self.V

        # Start with the first vertex
        key[0] = 0
        parent[0] = -1  # First node is always root of MST

        # The MST will have V vertices
        for _ in range(self.V):

            # 1. Pick the minimum key vertex not yet in MST
            u = self.min_key(key, mst_set)

            # 2. Add the picked vertex to the MST set
            mst_set[u] = True

            # 3. Update key and parent of adjacent vertices
            for v in range(self.V):
                # self.graph[u][v] is the edge weight
                # We update 'v' only if:
                # - An edge exists (weight > 0)
                # - 'v' is not already in the MST
                # - The edge weight is smaller than the current key[v]
                if (self.graph[u][v] > 0 and
                        not mst_set[v] and
                        key[v] > self.graph[u][v]):

                    key[v] = self.graph[u][v]
                    parent[v] = u

        # After the loop, the MST is stored in 'parent'
        self.print_mst(parent)

# --- Main execution ---
if __name__ == "__main__":
    # Create a graph with 5 vertices
    g = Graph(5)

    # Define the graph as an adjacency matrix (weights)
    # 0 means no direct edge
    g.graph = [
        [0, 2, 0, 6, 0],
        [2, 0, 3, 8, 5],
        [0, 3, 0, 0, 7],
        [6, 8, 0, 0, 9],
        [0, 5, 7, 9, 0]
    ]

    print("--- Prim's Algorithm for MST ---")
    print("Graph (Adjacency Matrix):")
    for row in g.graph:
        print(row)
    print("\nFinding MST...")

    # Run Prim's algorithm
    g.prim_mst()
```

```
def knapsack_01(values, weights, W):
    """
    Solves the 0-1 knapsack problem using dynamic programming.

    Args:
        values (list): A list of values for each item.
        weights (list): A list of weights for each item.
        W (int): The maximum capacity of the knapsack.

    Returns:
        tuple: A tuple containing:
               (int) The maximum value that can be achieved.
               (list) A list of indices of the items included.
    """
    n = len(values)

    # Create a DP table
    # dp[i][w] will store the maximum value using
    # items 1..i with a maximum weight of w
    dp = [[0 for _ in range(W + 1)] for _ in range(n + 1)]

    # Build the table dp[][] in a bottom-up manner
    for i in range(1, n + 1):  # Iterate over items
        for w in range(1, W + 1):  # Iterate over capacities

            # Get weight and value for the current item
            # (i-1 because lists are 0-indexed)
            current_weight = weights[i - 1]
            current_value = values[i - 1]

            # Case 1: If the current item's weight is more than
            # the current capacity 'w', we can't include it.
            if current_weight > w:
                dp[i][w] = dp[i - 1][w]
            else:
                # Case 2: We have two choices:
                # 1. Don't include the item: Value is dp[i-1][w]
                # 2. Include the item: Value is current_value +
                #    dp[i-1][w - current_weight]

                # We take the maximum of these two choices
                dp[i][w] = max(
                    dp[i - 1][w],  # Don't include
                    current_value + dp[i - 1][w - current_weight] # Include
                )

    # The maximum value is in the bottom-right corner
    max_value = dp[n][W]

    # --- Backtrack to find which items were included ---
    included_items = []
    res = max_value
    w = W
    for i in range(n, 0, -1):
        if res <= 0:
            break
        # If res is the same as the row above, it means
        # item 'i' was not included.
        if res == dp[i - 1][w]:
            continue
        else:
            # This item (i-1) was included
            included_items.append(i - 1)

            # Subtract its value and weight
            res = res - values[i - 1]
            w = w - weights[i - 1]

    included_items.reverse() # Show in original order

    return max_value, included_items

# --- Main execution ---
if __name__ == "__main__":
    values = [60, 100, 120]
    weights = [10, 20, 30]
    W = 50  # Knapsack capacity

    print("--- 0-1 Knapsack Problem ---")
    print(f"Item Values:  {values}")
    print(f"Item Weights: {weights}")
    print(f"Knapsack Capacity: {W}\n")

    max_val, items = knapsack_01(values, weights, W)

    print(f"Maximum value in knapsack: {max_val}")
    print(f"Item indices included: {items}")

    print("\nItems included (Weight, Value):")
    for i in items:
        print(f"  - Item {i} (Weight: {weights[i]}, Value: {values[i]})")

    # Example 2
    values_2 = [10, 40, 30, 50]
    weights_2 = [5, 4, 6, 3]
    W_2 = 10

    print("\n--- Example 2 ---")
    print(f"Item Values:  {values_2}")
    print(f"Item Weights: {weights_2}")
    print(f"Knapsack Capacity: {W_2}\n")

    max_val_2, items_2 = knapsack_01(values_2, weights_2, W_2)

    print(f"Maximum value in knapsack: {max_val_2}")
    print(f"Item indices included: {items_2}")
```

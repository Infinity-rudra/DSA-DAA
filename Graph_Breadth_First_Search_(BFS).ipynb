```
from collections import deque

def bfs(graph, start_node):
    """
    Performs Breadth-First Search on a graph and prints the visited nodes.

    Args:
        graph (dict): The graph represented as an adjacency list.
        start_node: The node to start the traversal from.
    """

    # A set to keep track of visited nodes to avoid cycles
    visited = set()

    # A queue for BFS
    queue = deque()

    # Check if the start node is valid
    if start_node not in graph:
        print(f"Error: Start node '{start_node}' not in graph.")
        return

    # Mark the start node as visited and enqueue it
    visited.add(start_node)
    queue.append(start_node)

    print("BFS Traversal Order:")

    while queue:
        # Dequeue a vertex from the front of the queue
        current_node = queue.popleft()
        print(current_node, end=" ")

        # Get all adjacent vertices of the dequeued vertex
        # If a neighbor has not been visited, mark it as visited
        # and enqueue it.
        for neighbor in graph[current_node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)
    print() # For a newline at the end

# --- Main execution ---
if __name__ == "__main__":
    # Example graph represented as an adjacency list (dictionary)
    # This graph is unconnected, with components {A, B, C, D, E} and {F}
    graph = {
        'A': ['B', 'C'],
        'B': ['A', 'D', 'E'],
        'C': ['A', 'E'],
        'D': ['B', 'E'],
        'E': ['B', 'C', 'D'],
        'F': []
    }

    print("--- Example 1 ---")
    print(f"Graph: {graph}")
    # Start BFS from node 'A'
    bfs(graph, 'A')

    print("\n--- Example 2 ---")
    # Start BFS from node 'D'
    bfs(graph, 'D')

    print("\n--- Example 3 (Different Graph) ---")
    graph_2 = {
        '1': ['2', '3'],
        '2': ['1', '4', '5'],
        '3': ['1', '5'],
        '4': ['2'],
        '5': ['2', '3']
    }
    print(f"Graph: {graph_2}")
    bfs(graph_2, '1')
```

```
def dfs_recursive(graph, node, visited):
    """
    Recursive helper function for DFS.
    """
    # Mark the current node as visited and print it
    visited.add(node)
    print(node, end=" ")

    # Recur for all the vertices adjacent to this vertex
    for neighbor in graph[node]:
        if neighbor not in visited:
            dfs_recursive(graph, neighbor, visited)

def dfs(graph, start_node):
    """
    Performs Depth-First Search on a graph starting from a given node.

    Args:
        graph (dict): The graph represented as an adjacency list.
        start_node: The node to start the traversal from.
    """

    # A set to keep track of visited nodes to avoid cycles
    visited = set()

    # Check if the start node is valid
    if start_node not in graph:
        print(f"Error: Start node '{start_node}' not in graph.")
        return

    print("DFS Traversal Order:")
    # Call the recursive helper function
    dfs_recursive(graph, start_node, visited)
    print() # For a newline at the end

# --- Main execution ---
if __name__ == "__main__":
    # Example graph represented as an adjacency list (dictionary)
    # This graph is unconnected, with components {A, B, C, D, E} and {F}
    graph = {
        'A': ['B', 'C'],
        'B': ['A', 'D', 'E'],
        'C': ['A', 'E'],
        'D': ['B', 'E'],
        'E': ['B', 'C', 'D'],
        'F': []
    }

    print("--- Example 1 ---")
    print(f"Graph: {graph}")
    # Start DFS from node 'A'
    dfs(graph, 'A')

    print("\n--- Example 2 ---")
    # Start DFS from node 'D'
    dfs(graph, 'D')

    print("\n--- Example 3 (Different Graph) ---")
    graph_2 = {
        '1': ['2', '3'],
        '2': ['1', '4', '5'],
        '3': ['1', '5'],
        '4': ['2'],
        '5': ['2', '3']
    }
    print(f"Graph: {graph_2}")
    dfs(graph_2, '1')
```
